```python
def print_args(*args):
    for arg in args:
        print(arg)

print_args('apple','orange',1,10,'mango','peacock')

print("")

def print_kwargs(**kwargs):
    for key,value in kwargs.items():
        print(f"{key} : {value}")

print_kwargs(name = "Alias" , age = 22 )


def args_kwargs(*args,**kwargs):
    print("args : ",args)
    print("kwargs : ",kwargs)

args_kwargs('apple','orange',name='alias',age=30)
```
