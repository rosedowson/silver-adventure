```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
```

```python
def softmax(x):
    return np.exp(x)/np.sum(np.exp(x),axis=0)

x=np.array([1,2,3,5,9],dtype=np.float32)
output=softmax(x)
print(output)


x=torch.tensor([1,2,3,5],dtype=torch.float32)
output=torch.softmax(x,dim=0)
print(output)
```

```python
def cross_entropy(y,y_predicted):
    return -1/len(y)*np.sum(y*np.log(y_predicted))

y=[1,0,0]
y_predicted=[0.8,0.1,0.1]
output=cross_entropy(y,y_predicted)
print(output)

y=torch.tensor([1],dtype=torch.long)
y_predicted=torch.tensor([[0.8,0.1,0.1]])

loss=nn.CrossEntropyLoss()
l=loss(y_predicted,y)
print(l.item())
```
