```python
x=np.array([1,2,2,6],dtype=np.float32)
y=x=np.array([1,3,2,8],dtype=np.float32)
w=0

# forward
def forward(x):
    return x*w

# loss
def loss(y,y_hat):
    return ((y-y_hat)**2).mean()
# gradient
# MSE = 1/n*(y-w*x)**2
# dJ/dw=2/n*(y-w*x)*x
def gradient(x,y,y_hat):
    return (-2*np.dot(x,y-y_hat)/len(x)).mean()

print(f"Prediction before training : f(5) = {forward(5):.3f}")

# training
# learning rate
lr=0.001

# iterations
# w = w-lr*dJ/dW
for epoch in range(300):
    # prediction
    y_hat=forward(x)
    # loss
    l = loss(y,y_hat)
    #dw
    dw = gradient(x,y,y_hat)
    w = w - lr * dw

    if epoch % 50 == 0:
        print(f'epoch {epoch} : w = {w:.3f} ,loss = {l:.8f}')

print(f"Prediction after training : f(5) = {forward(5):.3f}")
```

```python
x = torch.tensor([1,2,3,4],dtype=torch.float32)
y = torch.tensor([3,2,1,6],dtype=torch.float32)
w = torch.tensor(0.,dtype=torch.float32,requires_grad=True)

# forward 
def forward(x):
    return w*x

print(f'Before training f(5) : {forward(5):.3f}')

# training
learning_rate = 0.001

loss = nn.MSELoss()
optimizer = torch.optim.SGD([w],lr=learning_rate)

for epoch in range(1000):
    # forward pass
    y_predict=forward(x)
    # loss
    l=loss(y,y_predict)
    l.backward()
    with torch.no_grad():
        w -= learning_rate * w.grad
    w.grad.zero_()
    if epoch % 100 == 0:
        print(f'epoch {epoch} : w = {w:.3f} , loss = {l:.8f}')

print(f'After training f(5) : {forward(5):.3f}')

```

```python
x = torch.tensor([1,2,3,4],dtype=torch.float32)
y = torch.tensor([3,2,1,5],dtype=torch.float32)
w = torch.tensor(0.,dtype=torch.float32,requires_grad=True)

# forward 
def forward(x):
    return w*x

print(f'Before training f(5) : {forward(5):.3f}')

# training
learning_rate = 0.001

loss = nn.MSELoss()
optimizer = torch.optim.SGD([w],lr=learning_rate)

for epoch in range(1000):
    # forward pass
    y_predict=forward(x)
    # loss
    l=loss(y,y_predict)
    l.backward()
    optimizer.step()
    optimizer.zero_grad()
    if epoch % 100 == 0:
        print(f'epoch {epoch} : w = {w:.3f} , loss = {l:.8f}')

print(f'After training f(5) : {forward(5):.3f}')

```
